####**TASK P18.2** <font color= orange>**Complete Rosalind Problem #5  Dictionaries**</font>

````python
# Enter code for Rosalind Problem #5 here:

# Initialize an empty dictionary to store word counts
word_counts = {}

# Open the input file
with open('/content/rosalind_ini6.txt', 'r') as file:
    # Read the entire content of the file
    content = file.read()

    # Split the content into words based on spaces
    words = content.split()

    # Iterate through each word and count its occurrences
    for word in words:
        # Increment the count for the current word, or set to 1 if new
        word_counts[word] = word_counts.get(word, 0) + 1

# Print the word counts in the specified format (key value pairs)
for word, count in word_counts.items():
    print(f"{word} {count}")
````

#<font color=blue>**P19.**</font> ```count()``` – counting things made easy

####**TASK P19.1** <font color= orange>**Complete Rosalind Problem #6  Counting DNA Nucleotides**</font>

````python
# Enter code for Rosalind Problem #6 here:

# Initialize counters for each nucleotide
counts = {'A': 0, 'C': 0, 'G': 0, 'T': 0}

# Open the input file containing the DNA string
with open('/content/rosalind_dna.txt', 'r') as file:
    dna_string = file.read().strip() # Read the DNA string and remove any trailing newline

# Iterate through the DNA string and count each nucleotide
for nucleotide in dna_string:
    if nucleotide in counts:
        counts[nucleotide] += 1

# Print the counts in the specified format
print(f"{counts['A']} {counts['C']} {counts['G']} {counts['T']}")
````

####**TASK P19.2** Modify your code for the Rosalind #5 Dictionary problem so that it uses count( ) to count the words.

````python
# Enter code for Rosalind Problem #5 here:

# Open the input file
with open('/content/rosalind_ini6.txt', 'r') as file:
    content = file.read()

# Split the content into words
words = content.split()

# Create a dictionary
word_counts = {}

# Count each word using count()
for word in words:
    word_counts[word] = words.count(word)

# Print the word counts
for word, count in word_counts.items():
    print(f"{word} {count}")
````
